```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Dataset path
DATASET_PATH = "../dataset/fir_frames.npz"

# Load dataset
data = np.load(DATASET_PATH, allow_pickle=True)

# Display available arrays
print("Available arrays:")
print(data.files)
```

```python
X = data["X"]
y = data["y"]
sessions = data["session"]
file_ids = data["file_id"]
classes = data["classes"]

print("========== DATASET INFORMATION ==========")

print("X shape:", X.shape)
print("X dtype:", X.dtype)

print("y shape:", y.shape)
print("y dtype:", y.dtype)

print("Sessions shape:", sessions.shape)
print("File IDs shape:", file_ids.shape)

print("\nClasses:")
print(classes)

print("\nUnique sessions:")
print(np.unique(sessions))

print("\nUnique labels:")
print(np.unique(y))
```

```python
print("Total frames:", len(X))

print("Labelled frames:", np.sum(y != -1))

print("Unlabelled frames:", np.sum(y == -1))

print("Number of sessions:", len(np.unique(sessions)))
```
